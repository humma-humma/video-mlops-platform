```python
from pathlib import Path

import cv2

import matplotlib.pyplot as plt
```

```python
video_path = Path("data/inputs/videos")
video_path.is_dir()
```

```python
frame_durations = []
frame_counts = []

for video_file in video_path.glob("*.mp4"):
    cap = cv2.VideoCapture(str(video_file))

    fps = cap.get(cv2.CAP_PROP_FPS)
    frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
    duration = frame_count / fps

    frame_durations.append(duration)
    frame_counts.append(frame_count)

    cap.release()
```

```python
plt.hist(frame_durations, bins="auto")
plt.title("Histogram of Video Durations (s)")
plt.xlabel("Duration (s)")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.hist(frame_counts, bins="auto")
plt.title("Histogram of Frame Counts")
plt.xlabel("Frame Count")
plt.ylabel("Frequency")
plt.show()
```
